```r
library(ggplot2)
```

```r
# data
n <- 10 # number of trials
p <- 0.7 # probability of success


vaccine_df <- data.frame(
    vaccinated = 0:n,
    probability = dbinom(0:n, size = n, prob = p)
)

ggplot(vaccine_df, aes(x = vaccinated, y = probability)) +
    geom_bar(stat = "identity", fill = "steelblue") +
    labs(
        title = "Binomial Distribution of Vaccinated Individuals",
        x = "Number of Vaccinated Individuals",
        y = "Probability"
    ) +
    theme_minimal()
```

```r
lambda <- 4 # average number of extreme weather events per year
x <- 0:15
y <- dpois(x, lambda)
poisson_df <- data.frame(events = x, probability = y)

ggplot(poisson_df, aes(x = events, y = probability)) +
    geom_bar(stat = "identity", fill = "firebrick") +
    labs(
        title = "Poisson Distribution of Extreme Weather Events",
        x = "Number of Extreme Weather Events",
        y = "Probability"
    ) +
    theme_minimal()
```

```r
# average daily calories intake
mean <- 2300
sd <- 400
x <- seq(1000, 3500, length.out = 200)
y <- dnorm(x, mean = mean, sd = sd)
calories_df <- data.frame(calories = x, density = y)

ggplot(calories_df, aes(x = calories, y = density)) +
    geom_line(color = "darkgreen", linewidth = 1.2) +
    labs(
        title = "Normal Distribution of Daily Calorie Intake",
        x = "Daily Calorie Intake",
        y = "Density"
    ) +
    theme_minimal()
```
