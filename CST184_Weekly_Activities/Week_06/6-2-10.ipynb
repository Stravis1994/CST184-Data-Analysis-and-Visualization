```python
import pandas as pd

df = pd.read_csv("week6_grouping_filtering.csv")
print(df.head())
```

```python
# Check for missing values
print(df.isnull().sum())

# Drop rows with missing attendance
df_cleaned = df.dropna(subset=['Attendance'])
print("Rows before cleaning:", len(df))
print("Rows after cleaning:", len(df_cleaned))
print("Rows dropped:", len(df) - len(df_cleaned))

# 5 rows were dropped due to missing attendance values. this lets us know that we have a complete dataset for our analysis.
```

```python
df_2023 = df_cleaned[df_cleaned['Year'] == 2023]
print("Rows in 2023 dataset:", len(df_2023))
df_2023.head()

# there are 58 rows of data for the year 2023.
```

```python
summary = df_2023.groupby(['Region', 'Gender'])['Attendance'].agg(['mean', 'count']).reset_index()
print(summary)
```

```python
pivot = df_2023.pivot_table(values='Attendance', index='Region', columns='Gender', aggfunc='mean')
print(pivot)
```

```python
region_avg = df_2023.groupby('Region')['Attendance'].mean()
high_regions = region_avg[region_avg > 75]
print(high_regions)

#based on the regional averages needing to be above 85 for this we get no results if we lower the treshold to 75 we will see that the West region as the highest average attendance followed by the South region.
```

```python
result = (
    df_cleaned[df_cleaned['Year'] == 2023]
    .groupby(['Region', 'Gender'])['Attendance']
    .agg(['mean', 'count'])
    .reset_index()
)
print(result)
```
