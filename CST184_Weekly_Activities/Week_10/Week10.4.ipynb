```r
health_df <- read.csv("sdg3_sugar_obesity_data.csv")
library(ggplot2)
ggplot(health_df, aes(x = Sugar_Intake_g_per_day, y = Obesity_Rate_percent)) +
  geom_point(color = "darkgreen") +
  geom_smooth(method = "lm", se = TRUE, color = "red") +
  labs(
    title = "Regression: Sugar Intake vs Obesity Rate",
    x = "Daily Sugar Intake (g)",
    y = "Obesity Rate (%)"
  ) +
  theme_minimal()
```
