```r
# Simulation settings
supply_mean <- 25
supply_sd <- 5
demand_mean <- 40
demand_sd <- 15
```

```r
results <- replicate(1000, {
  supply <- rnorm(30, mean = supply_mean, sd = supply_sd)
  demand <- rnorm(30, mean = demand_mean, sd = demand_sd)
  all(supply >= demand)
})
all(results)
```

```r
results <- replicate(1000, {
  supply <- rnorm(30, mean = supply_mean, sd = supply_sd)
  demand <- rnorm(30, mean = demand_mean, sd = demand_sd)
  mean(supply >= demand)
})
mean(results)
```
