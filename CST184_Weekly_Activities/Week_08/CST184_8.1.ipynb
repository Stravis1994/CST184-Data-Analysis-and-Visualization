```r
a <- matrix(c(1, 2, 3, 4), nrow=2)
b <- matrix(c(2, 0, 1, 3), nrow=2)

print(a)
print(b)

result <- a %*% b
print(result)
```

```r
# changing one value in b changed the result of the matrix multiplication in col 1, col 2 is the same as before, but col 1 is different because of the change in b[1,2] from 0 to 4
a <- matrix(c(1, 2, 3, 4), nrow=2)
b <- matrix(c(2, 4, 1, 3), nrow=2)

result <- a %*% b
print(result)
```
